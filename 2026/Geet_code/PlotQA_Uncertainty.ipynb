```python
from transformers import AutoModelForImageTextToText, AutoProcessor
from lm_polygraph.model_adapters.visual_whitebox_model import VisualWhiteboxModel
from lm_polygraph import estimate_uncertainty
from lm_polygraph.estimators import MaximumTokenProbability, MaximumSequenceProbability, SemanticEntropy, EigValLaplacian
```

```python
base_model = AutoModelForImageTextToText.from_pretrained("microsoft/kosmos-2-patch14-224")
processor = AutoProcessor.from_pretrained("microsoft/kosmos-2-patch14-224")

# Create whitebox model
model = VisualWhiteboxModel(base_model, processor)
```

```python
def uncertainty_battery(model, input_text, input_image):
    estimators = [
        MaximumTokenProbability(),
        MaximumSequenceProbability(),
        SemanticEntropy(),
        EigValLaplacian()
    ]

    list_uncertainty = [estimate_uncertainty(model, estimator, input_text=input_text, input_image=input_image) for estimator in estimators]

    return list_uncertainty
```

```python
from lm_polygraph.estimators import MeanTokenEntropy
```

```python
def mte_estimator(model, input_text, input_image):
    messages = [
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "image": input_image,
                },
                {"type": "text", "text": input_text},
            ],
        }
    ]

    processed_text = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )

    estimator = MeanTokenEntropy()
    response = estimate_uncertainty(model, estimator, input_text=processed_text, input_image=input_image)
    return response
```

```python
def msp_estimator(model, input_text, input_image):
    messages = [
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "image": input_image,
                },
                {"type": "text", "text": input_text},
            ],
        }
    ]

    processed_text = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )

    estimator = MaximumSequenceProbability()
    response = estimate_uncertainty(model, estimator, input_text=processed_text, input_image=input_image)
    return response
```

```python
msp_responses = []

for row in df_sample.itertuples():
    print(f"Processing row {row.Index}...")

    input_string = row.question_string
    image_index = row.image_index
    input_image = image_index_to_path(image_index)
    ytrue = row.answer

    response = msp_estimator(model, input_string, input_image)
    print(response)
    
    msp_responses.append(response)
```

```python
msp_scores = np.array([response.uncertainty for response in msp_responses])
msp_scores
```

```python
uncertainty_scores = -msp_scores
uncertainty_scores  
```

```python
percentage_errors = np.abs((ytrue - ypred) / (ytrue + np.finfo(float).eps)) * 100
percentage_errors
```

```python
from scipy.stats import pearsonr

pearson_corr, p_value = pearsonr(percentage_errors, uncertainty_scores)
print(f"Pearson Correlation: {pearson_corr}, P-value: {p_value}")

slope, intercept = np.polyfit(percentage_errors, uncertainty_scores, 1)
plt.scatter(percentage_errors, uncertainty_scores)
plt.plot(percentage_errors, slope * percentage_errors + intercept, color='red')
plt.xlabel("Percentage Error")
plt.ylabel("Uncertainty Score (Negative MSP)")
plt.title("Uncertainty Score (Negative MSP) vs Percentage Error")
plt.show()
```
